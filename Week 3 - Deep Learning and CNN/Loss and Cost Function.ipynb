# Loss function
The derivative of loss function with respect to weight is used to update the weight using the formula: New Weight = Old Weight − (Learning Rate × Derivative)

Different types of loss functions:

cost function = sum(yi - yi')

loss function or mean error = (1/n)(sum(yi - yi'))

mean square error = (1/n)(sum(yi - yi')^2)

log loss or binary cross entropy = -(1/n)( sum(yi(log(yi')) - (1 - yi)(log(1 - yi)) )

```python
import numpy as np
```

```python
y_predicted = np.array([1,1,0,0,1])
y_true = np.array([0.30,0.7,1,0,0.5])
```

```python
def mae(y_predicted, y_true):
    total_error = 0
    for yp, yt in zip(y_predicted, y_true):
        total_error += abs(yp - yt)
    print("Total error is:",total_error)
    mae = total_error/len(y_predicted)
    print("Mean absolute error is:",mae)
    return mae
```

```python
mae(y_predicted, y_true)
```

```python
def log_loss(y_true, y_predicted):
    y_predicted_new = [max(i,epsilon) for i in y_predicted]
    y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new]
    y_predicted_new = np.array(y_predicted_new)
    return -np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))
```

```python
log_loss(y_true, y_predicted)
```
