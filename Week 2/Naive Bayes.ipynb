# Naive Bayes
classification algorithm based on Bayes' Theorem of Probability and the assumption that features are conditionally independent given the class.

GaussianNB = Used when features are continuous numerical values. We calculate the mean for each outcome then calculate the probability of the input for each output scenario and choose the highest one.

MultinomialNB = Used when features represent counts or frequencies. eg: the number of words is calculated to detect spam mails.

```python
import pandas as pd
```

```python
df = pd.read_csv("data/titanic.csv")
df.head()
```

```python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head()
```

```python
inputs = df.drop('Survived',axis='columns')
target = df.Survived
```

```python
dummies = pd.get_dummies(inputs.Sex)
dummies.head(3)
```

```python
inputs = pd.concat([inputs,dummies],axis='columns')
inputs.head(3)
```

```python
inputs.drop(['Sex','male'],axis='columns',inplace=True)
inputs.head(3)
```

```python
inputs.columns[inputs.isna().any()]
```

```python
inputs.Age[:10]
```

```python
inputs.Age = inputs.Age.fillna(inputs.Age.mean())
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(inputs,target,test_size=0.3)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
X_test[0:10]
```

```python
y_test[0:10]
```

```python
model.predict(X_test[0:10])
```

```python
model.predict_proba(X_test[:10])
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(GaussianNB(),X_train, y_train, cv=5)
```

```python
df1 = pd.read_csv("data/spam.csv")
df1.head()
```

```python
df1.groupby('Category').describe()
```

```python
df1['spam']=df1['Category'].apply(lambda x: 1 if x=='spam' else 0)
df1.head()
```

```python
from sklearn.model_selection import train_test_split
X1_train, X1_test, y1_train, y1_test = train_test_split(df1.Message,df1.spam)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
v = CountVectorizer()
X1_train_count = v.fit_transform(X1_train.values)
X1_train_count.toarray()[:2]
```

```python
from sklearn.naive_bayes import MultinomialNB
model1 = MultinomialNB()
model1.fit(X1_train_count,y1_train)
```

```python
emails = [
    'Hey mohan, can we get together to watch footbal game tomorrow?',
    'Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!'
]
emails_count = v.transform(emails)
model1.predict(emails_count)
```

```python
X1_test_count = v.transform(X1_test)
model1.score(X1_test_count, y1_test)
```

```python
from sklearn.pipeline import Pipeline
clf = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('nb', MultinomialNB())
])
```

Pipeline - instead of doing it in two steps pipline makes it into one

```python
clf.fit(X1_train, y1_train)
```

```python
clf.score(X1_test,y1_test)
```

```python
clf.predict(emails)
```
