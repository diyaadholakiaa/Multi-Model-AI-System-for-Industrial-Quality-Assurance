# Principal Component Analysis (PCA)
To train models faster we have to reduce the amount of data to train it on. Sometimes there are useless datapoints in the data which can be easily removed. PCA identifies principal components which can be used which have the maximum relevance to the model nad removes the rest. These components may or may not be the same as the original ones.

```python
from sklearn.datasets import load_digits
import pandas as pd

dataset = load_digits()
dataset.keys()
```

```python
df = pd.DataFrame(dataset.data, columns=dataset.feature_names)
df.head()
```

```python
dataset.target
```

```python
df.describe()
```

```python
X = df
y = dataset.target
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
X
```

```python
from sklearn.decomposition import PCA

pca = PCA(0.95)
X_pca = pca.fit_transform(X)
X_pca.shape
```

pca such that 95% of variance is retained

```python
pca.explained_variance_ratio_
```

```python
pca.n_components_
```

PCA created 29 components out of original 64 columns

```python
X_pca
```

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_pca, y_train)
model.score(X_test_pca, y_test)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
X_pca.shape
```

PCA with only two components

```python
X_pca
```

```python
pca.explained_variance_ratio_
```

both combined retains 0.14+0.13=0.27 or 27% of important feature information

```python
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_pca, y_train)
model.score(X_test_pca, y_test)
```
