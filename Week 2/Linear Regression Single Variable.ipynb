```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

pandas - for organising and cleaning data

numpy - for fast numerical computation

matplotlib.pyplot - for visualising data

sklearn (scikit-learn) - for machine learning, to classify/ predict/ finding groups or clusters in data

```python
df = pd.read_csv("data/canada_per_capita_income.csv")
df
```

```python
%matplotlib inline
plt.xlabel("Years")
plt.ylabel("Income")
plt.scatter(df.year, df['per capita income (US$)'], color='blue', marker='o')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['year']], df[['per capita income (US$)']])
```

The double bracket is used to make the series into dataframe, else is shows error. Also if the title of the column has () then df.title, does not work you have to use df[]

```python
reg.predict([[2020]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
828.465075*2020-1632210.75785546
```

if you have a csv file that has different years, then you can do d=pd.read_csv() p=reg.predict(d) to predict price for all those values. then by d['prices'] you can add the prices of those years to d. Then d.to_csv("", index = False) to turn it into a csv file

```python
plt.scatter(df.year, df['per capita income (US$)'], color='blue', marker='o')
plt.plot(df.year, reg.predict(df[['year']]), color='red')
```

this is the best fit line

```python
import pickle
```

```python
with open('model_linear_regression_pickle', 'wb') as f:
    pickle.dump(reg, f)
```

open() - creates file named model_linear_regression in wb or write binary mode. dump is the step while saves the model as a file in your computer

```python
with open('model_linear_regression', 'rb') as f:
    mp = pickle.load(f)
```

rb - read binary

```python
mp.predict([[2021]])
```
