# Logistic Regression Binary
Used for prediction when only two fixed outcomes are possible. Uses a sigmoid function. We find the predicted value from linear regression and put it in the sigmoid function

```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("data/insurance_data.csv")
df.head()
```

```python
plt.scatter(df.age,df.bought_insurance,marker='+',color='red')
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['age']],df.bought_insurance,train_size=0.8)
```

```python
X_test
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
X_test
```

```python
y_predicted = model.predict(X_test)
```

```python
model.predict_proba(X_test)
```

```python
model.score(X_test,y_test)
```

```python
y_predicted
```

```python
X_test
```

```python
model.coef_
```

```python
model.intercept_
```

```python
import math
def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

```python
def prediction_function(age):
    z = model.coef_ * age - model.intercept_
    y = sigmoid(z)
    return y
```

```python
age = 35
prediction_function(age)
```

```python
age = 43
prediction_function(age)
```
